```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
```

Basics

```python
X=np.linspace(0,5,11)
y=X**2
plt.plot(X,y)
plt.title("Hello")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
```

Subplots in Matplotlib

```python
plt.subplot(2,2,1)
plt.plot(X,y)
plt.subplot(2,2,2)
plt.plot(X**2,X**2)
plt.subplot(2,2,3)
plt.plot(y,X)
plt.subplot(2,2,4)
plt.plot(X,y)
```

Object Oriented way

```python
fig=plt.figure(figsize=(6,8),dpi=100)
axis1=fig.add_axes([0,0,1,1])
axis1=plt.plot(X,y)
axis2=fig.add_axes([.4,.5,0.3,0.3])
axis2=plt.plot(y,X)
```

Types of plots

```python
plt.scatter(X,y)
```

```python
plt.hist(X,y)
```

```python
plt.boxplot(X)
```

Saving plots

```python
fig=plt.figure()
ax=fig.add_axes([0,0,1,1])
ax=plt.plot(X,y)
plt.savefig("plot")
```
