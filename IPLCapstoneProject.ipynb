Loading the libraries and dataset

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
df=pd.read_csv('IPL.csv')
df.head()
```

Basic information

```python
df.info()
```

Check the size of rows and columns of the dataset

```python
print(f"Your rows are {df.shape[0]} and your columns are {df.shape[1]}")
```

Now let's see how many columns have null values in total

```python
df.isna().sum()
```

Which team won the most matches?

```python
match_wins=df['match_winner'].value_counts()
sns.barplot(y=match_wins.index,x=match_wins.values,palette='viridis')
plt.title("Most match by team")
```

Toss Decision trends

```python
sns.countplot(x=df['toss_decision'])
plt.title("Toss Decision Trends")
```

Toss winner vs match winner

```python
count=df[df['toss_winner']==df['match_winner']]['match_id'].count()
percentage=(count*100)/df.shape[0]
percentage.round(2)
```

How do teams win? (Runs or Wickets)

```python
sns.countplot(x=df['won_by'])
plt.title("Won by")
```

Most "Player of the match" Awards

```python
count =df['player_of_the_match'].value_counts().head(10)
sns.barplot(x=count.values,y=count.index,palette='mako')
plt.title("Top 10 playres with man of the match")
```

Tow top scorers

```python
highScorer=df.groupby('top_scorer')['highscore'].sum().sort_values(ascending=False).head(2)
highScorer.plot(kind="bar")
```

10 Best bowling figures

```python
df['highestWickets']=df['best_bowling_figure'].apply(lambda x:x.split('--')[0])
df['highestWickets']=df['highestWickets'].astype(int)
bestBowlers=df.groupby('best_bowling')['highestWickets'].sum().sort_values(ascending=False).head(10)
bestBowlers.plot(kind='barh')
```

Venue Analysis

Most matches played by Venue

```python
venueCount=df['venue'].value_counts()
sns.barplot(y=venueCount.index,x=venueCount.values)
```

Custom Questions & Insights

Who won the highest margin by runs?

```python
df[df['won_by']=='Runs'].sort_values(by='margin',ascending=False).head(1)[['match_winner','margin']]
```

Which player has highest individual score

```python
df[df['highscore']==df['highscore'].max()][['top_scorer','highscore']]
```

Which bowler had the best bowling figures

```python
df[df['highestWickets']==df['highestWickets'].max()][['best_bowling','best_bowling_figure']]
```
