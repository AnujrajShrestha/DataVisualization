```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data=pd.read_csv('price_quantity.csv')
df=pd.DataFrame(data)
df
```

Calculations

```python
df['TR']=df["Quantity"]*df['Price']
df['AR']=(df['TR']/df['Quantity']).astype(int)
df['MR']=df['TR'].diff()/df["Quantity"].diff()
df
```

Ploting

```python
plt.figure(figsize=(10,6))
sns.lineplot(data=df,x="Quantity",y="TR",marker='o',label="TR")
sns.lineplot(data=df,x="Quantity",y="AR",marker='o',label="AR")
sns.lineplot(data=df,x="Quantity",y="MR",marker='o',label="MR")
plt.grid(True)
plt.title("TR, AR and MR Curves")
plt.xlabel("Quantity")
plt.ylabel("Revenue")
plt.legend()
plt.show()
```
