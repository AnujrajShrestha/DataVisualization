```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data=pd.read_csv("quantity_tfc_tvc.csv")
df=pd.DataFrame(data)
df
```

Calculations

```python
df['TC']=df['TFC']+df['TVC']
df['AFC']=(df['TFC']/df['Q']).round(2)
df['AVC']=(df['TVC']/df['Q']).round(2)
df['MC']=df['TC'].diff()/df['Q'].diff()
df
```

ploting

```python
print(df)
plt.figure(figsize=(10,6))
sns.lineplot(data=df,y='TFC',x='Q',marker='o',label="TFC")
sns.lineplot(data=df,y='TVC',x='Q',marker='o',label="TVC")
sns.lineplot(data=df,y='TC',x='Q',marker='o',label="TC")
sns.lineplot(data=df,y='AFC',x='Q',marker='o',label="AFC")
sns.lineplot(data=df,y='AVC',x='Q',marker='o',label="AVC")
sns.lineplot(data=df,y='MC',x='Q',marker='o',label="MC")
plt.xlabel("Quantity")
plt.ylabel("Cost")
plt.title("TFC,TVC,TC,AFC,AVC and MC curves")
plt.grid(True)
plt.legend()
plt.show()
```
